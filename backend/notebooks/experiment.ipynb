### Downlaod the dataset from gdrive

```python
import gdown
```

```python
url = 'https://drive.google.com/file/d/1FHDwF7gWgKMIlb0suMygK0sz1_niE8Gn/view?usp=sharing'
```

```python
file_id = url.split('/')[-2]
file_id
```

```python
gdown.download(prefix + file_id, 'chest_ct_scan.zip', quiet=False)
```
